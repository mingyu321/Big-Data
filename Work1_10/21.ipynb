T1-1. 데이터에서 IQR을 활용해 Fare컬럼의 이상치를 찾고, 이상치 데이터의 여성 수를 구하시오

```python
import pandas as pd
import numpy as np

df = pd.read_csv('/content/Titanic.csv')
df.info()
df.head()
```

```python
q1 = df['Fare'].quantile(.25)
q3 = df['Fare'].quantile(.25)
IQR = q3 - q1
```

```python
outdata_1 = df[ df['Fare'] < (q1-IQR*1.5)]
outdata_2 = df[ df['Fare'] > (q3+IQR*1.5)]

len(outdata_1), len(outdata_2)
```

```python
print( sum(outdata_1['Gender'] == 'female') + sum(outdata_2['Gender'] == 'female'))
```

T1-2. 주어진 데이터에서 이상치(소수점 나이)를 찾고 올림, 내림, 버림(절사)했을때 3가지 모두 이상치 'age' 평균을 구한 다음 모두 더하여 출력하시오

```python
import pandas as pd
import numpy as np

df = pd.read_csv('/content/basic1.csv')
```

```python
df.head()
```

```python
# 소수점 나이 추출/ 올림,내림,버림을 각각 적용했을 때의 평균 구하기/ 3가지 경우의 평균 모두 합
```

```python
df = df[ (df['age'] - np.floor(df['age'])) != 0 ] # 원본 age값 - 내림 age값의 계산 결과가 0이 아닌 경우 True 반환
df
```

```python
# 올림,내림,버림의 각 평균값
ceil_mean = np.ceil(df['age']).mean()
floor_mean = np.floor(df['age']).mean()
trunc_mean = np.trunc(df['age']).mean()
```

```python
print(ceil_mean + floor_mean + trunc_mean)
```

T1-3. 주어진 데이터에서 결측치가 80%이상 되는 컬럼은(변수는) 삭제하고, 80% 미만인 결측치가 있는 컬럼은 'city'별 중앙값으로 값을 대체하고 'f1'컬럼의 평균값을 출력하세요!

```python
# 결측치 처리(map 활용)
import pandas as pd
import numpy as np

df = pd.read_csv('/content/basic1.csv')
print(df.shape)
print(df.isna().sum())
df.head()
```

```python
df.isna().sum() / df.shape[0] # 결측치 비율: 결측치 수 / 전체 수
# df.shape[0]:전체 행의 수 반환  df.shape[1]:열
```

```python
df = df.drop(columns=['f3'])
df
```

```python
df['city'].value_counts()
```

```python
k_median = df[df['city'] == '경기']['f1'].median() # 'city'열의 값이 '경기'인 행을 필터링하고, 그중에서 'f1'열의 중앙값을 계산
s_median = df[df['city'] == '서울']['f1'].median()
d_median = df[df['city'] == '대구']['f1'].median()
b_median = df[df['city'] == '부산']['f1'].median()

k_median, s_median, d_median, b_median
```

```python
df.head()
```

```python
df['f1'] = df['f1'].fillna(df['city'].map({'경기':k_median, '서울':s_median, '대구':d_median, '부산':b_median}))
# map(): pandas에서 시리즈(Series)의 값들을 다른 값으로 매핑(mapping)하거나 변환
# df의 한 열은 시리즈 객체다.  df은 여러개의 시리즈가 같은 인덱스를 기준으로 모여 있는 2차원 구조

# 만약 f1 중앙값으로 대체한다면
# df['f1'] = df['f1'].fillna(df['f1'].median())
```

```python
print(df['f1'].mean())
```
